# 01 - Data Loading

## 1. Purpose
Load the raw monthly revenue extract and confirm it reads in cleanly with the
expected shape and columns before any transformation happens. This notebook
does not clean, validate, or analyze the data -- it only loads and inspects it.


## 2. Imports

```python
import pandas as pd

pd.set_option('display.max_rows', 30)
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')
```

## 3. Data Loading

```python
RAW_PATH = "../data/raw/monthly_revenue_raw.csv"

df = pd.read_csv(RAW_PATH)
print(f"Shape: {df.shape}")
df.head()
```

## 4. Analysis

```python
print(df.dtypes)
print()
print("Columns:", list(df.columns))
print()
print(df.tail())
```

## 5. Findings

- The raw extract contains **23 rows and exactly 2 columns**: `sales_month`
  (string, `YYYY-MM`) and `current_month_revenue` (integer).
- There is **no order-level, transaction-level, or customer-level data** in
  this extract -- no order IDs, no line items, no customer identifiers.
  This is a hard constraint on the rest of the project: order counts and
  Average Order Value (AOV) cannot be computed from this source (see
  `docs/assumptions.md`).
- The data appears to already be pre-aggregated to monthly grain upstream,
  not derived from raw transactions in this pipeline.


## 6. Conclusion

The raw file loads cleanly with the expected 2-column monthly-grain schema.
Proceed to `02_data_validation.ipynb` for formal data quality checks before
any metric calculation.
